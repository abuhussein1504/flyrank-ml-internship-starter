## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

---
1. The good research question is:

   > Which content pages should be reviewed first for refresh based on observable search and engagement signals?

2. The decision belongs to a **human**.

3. Action: The system simply tells the user to start with these pages.

4. Cost of a wrong recommendation:

   1. False Positive:

      > If it says "Review this page". But actually the page is fine -> The team wastes time.

   2. False Negative

      > The system ignores a page. But that page really needed updating -> The company may lose traffic.
